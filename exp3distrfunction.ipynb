```python
np.random.seed(42) 
lam = 3  
n = 1000
data = np.random.exponential(scale=1/lam, size=n)  
```

```python
mean_value = np.mean(data)
variance_value = np.var(data)
median_value = np.median(data)

print(f"Միջին արժեքը: {mean_value:.4f}")
print(f"Վարիացիան: {variance_value:.4f}")
print(f"Մեդիանը: {median_value:.4f}")
```

```python
sorted_data = np.sort(data)
cdf = np.arange(1, n + 1) / n

plt.figure(figsize=(8, 5))
plt.plot(sorted_data, cdf, label="Էմպիրիկ բաշխման ֆունկցիա (CDF)", color="blue")
```
